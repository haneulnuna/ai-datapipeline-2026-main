```python
data = {
    "order_date": [
        "2026-09-01", "2026-09-01", "2026-09-02", "2026-09-02",
        "2026-09-03", "2026-09-03", "2026-09-04", "2026-09-04",
        "2026-09-05", "2026-09-05"
    ],
    "customer": ["홍길동", "김철수", "이영희", "홍길동", "박민수", "김철수", "최지훈", "이영희", "홍길동", "박민수"],
    "category": ["주변기기", "주변기기", "컴퓨터", "저장장치", "주변기기", "저장장치", "컴퓨터", "주변기기", "저장장치", "컴퓨터"],
    "product": ["키보드", "마우스", "노트북", "USB", "마우스", "USB", "노트북", "키보드", "SSD", "모니터"],
    "quantity": [2, 1, 1, 5, 3, 4, 1, 2, 1, 1],
    "unit_price": [30000, 15000, 1200000, 12000, 15000, 12000, 1200000, 30000, 90000, 250000]
}
```

##### 막대그래프

- 항목별(카테고리) 크기를 비교할 때 사용

```python
df.groupby('category')['total_price'].sum()
```

```python
cate_sales = df.groupby('category')['total_price'].sum().reset_index()
```

```python
# x축

cate_sales['category']
```

```python
# y축

cate_sales['total_price']
```

```python
plt.figure(figsize=(10,5))

# 막대그래프 plt.bar()

plt.bar(x=cate_sales['category'], height=cate_sales['total_price'])
plt.title('항목별 판매금액', fontsize=15, fontweight='bold')
plt.xlabel('카테고리')
plt.ylabel('판매금액')

plt.ticklabel_format(style='plain', axis='y') # y축을 판매금액 그대로 출력. 2.5 1e6 과 같이 출력
plt.show
```

##### 선 그래프

- 날짜별 매출 계산

```python
df.groupby('order_date')['total_price'].sum().reset_index()
```

```python
daily_sales = df.groupby('order_date')['total_price'].sum().reset_index()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    daily_sales['order_date'],
    daily_sales['total_price'],
    marker='o',
    markersize=10,
    linewidth=2.5,
    color="#F5A47B"
)

plt.title('일자별 판매금액')
plt.ylabel('판매금액')
plt.ticklabel_format(style='plain', axis='y')

# plt.xticks(
#     daily_sales['order_date'],
#     daily_sales['order_date'].dt.strftime('%y%m%d')
# )

plt.show()
```

#### 히스토그램
- 표준편차, 분포와 관계있는 그래프
- 숫자 데이터의 분포를 확인할 때 사용

```python
plt.figure(figsize=(10, 5))

plt.hist(df['total_price'], bins=5)
plt.title('구매금액 분포도', fontsize=15)
plt.xlabel('구매금액 분포')
plt.ylabel('갯수')

plt.ticklabel_format(style='plain', axis='x')
plt.show()
```

#### 박스플롯
- 데이터의 이상치와 분포를 같이 확인할 때 사용

```python
plt.figure(figsize=(10, 5))

# 박스플롯
plt.boxplot(df['total_price'])
plt.title('구매금액 분포와 이상치', fontsize=15)
plt.ylabel('구매금액')

plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

##### 선 그래프 마커위치에 데이터 표시

```python
plt.figure(figsize=(10, 5))
plt.plot(daily_sales['order_date'], daily_sales['total_price'], 
         marker='o', 
         markersize=10, 
         linewidth=2.5, 
         color="#EF5B10")
# 마커위치에 실제 데이터 표시
for x, y in zip(daily_sales['order_date'], daily_sales['total_price']):
    plt.text(
        x, y,
        f"{y:,}",  # y값 = 금액
        ha='left',  # 가로축 정렬 left, center, right  - 정렬 이름 반대
        va='bottom'   # 세로축 정렬 top, center, bottom - 정렬 이름 반대
    )
plt.title('일자별 판매금액')
plt.ylabel('판매금액')
plt.ticklabel_format(style='plain', axis='y')
# 실제 데이터가 있는 위치의 x좌표 레이블 표시
plt.xticks(
    daily_sales['order_date'],
    daily_sales['order_date'].to_string('%y%m%d')
)
plt.show()
```

#### Seaborn Heatmap
- 숫자 데이터 분포를 색상으로 표현하는 그래프

```python
# 구매수량, 단가, 구매금액 간의 상관관계
df[['quantity', 'unit_price', 'total_price']].corr()
```

```python
corr = df[['quantity', 'unit_price', 'total_price']].corr()
```

```python
plt.figure(figsize=(10, 5))

# 히트맵
sns.heatmap(data=corr, annot=True, cmap='Blues')
plt.title('구매수량과 단가 간의 상관관계')
plt.show()
```
